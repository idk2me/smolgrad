```python
import numpy as np
from smolgrad import two_moons
import matplotlib.pyplot as plt
%matplotlib inline
```

```python
X_train, y_train, X_test, y_test = two_moons.get_training_split()
for x, y in zip(X_train[:50], y_train[:50]):
    print(f"input ({x[0].data:.4f}, {x[1].data:.4f}) -> label {y}")
```

```python
X_train_np = np.array([[x[0].data, x[1].data] for x in X_train])
X_test_np = np.array([[x[0].data, x[1].data] for x in X_test])

X_all = np.vstack([X_train_np, X_test_np])
y_all = np.concatenate([y_train, y_test])

plt.scatter(X_all[:, 0], X_all[:, 1], c=y_all)
```
